In scikit-learn, a ColumnTransformer is a powerful utility that allows you to apply different preprocessing transformations to specific columns of an array or pandas DataFrame within a single step

It solves a classic machine learning headache: handling mixed datasets (heterogeneous data) where numerical columns require scaling or imputation, while categorical columns require one-hot or ordinal encoding

# Why use a ColumnTransformer
- Prevents Data Leakage :  Centralizing preprocessing ensures your transformations are cleanly fit only on the training split and carried over identically to the test split.

- Cleaner Pipelines :  Instead of manually splitting, transforming, and gluing your dataframe columns back together, ColumnTransformer handles the concatenation automatically.

- Hyperparameter Tuning : It integrates seamlessly with scikit-learn Pipelines, allowing you to optimize your preprocessing choices (like scaling types or imputation strategies) using grid search.

```python
import numpy as np
import pandas as pd
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
df = pd.read_csv('covid_toy.csv')
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df.drop(columns = 'has_covid'), 
                                                    df['has_covid'],
                                                    test_size = 0.2, 
                                                    random_state = 42, 
                                                    stratify = df['has_covid'])
```

```python
from sklearn.compose import ColumnTransformer

transformer = ColumnTransformer(transformers = [
    ('tnf1', SimpleImputer(), ['fever']),
    ('tnf2', OrdinalEncoder(categories = [['Mild', 'Strong']]), ['cough']),
    ('tnf3', OneHotEncoder(sparse_output=False, drop = 'first'), ['gender','city'])
], remainder = 'passthrough')

X_train = transformer.fit_transform(X_train)
X_test = transformer.transform(X_test)
```
