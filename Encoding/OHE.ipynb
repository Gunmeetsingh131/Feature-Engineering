Columns with larger number of unique values are not suitable for One Hot Encoding. For example, the column 'brand' has 30 unique values. If we apply One Hot Encoding to this column, it will create 29 new columns (after dropping the first one). This can lead to a significant increase in the dimensionality of the dataset, which may not be desirable.

```python
counts = df['brand'].value_counts()
```

```python
df['brand'].nunique()
threshold = 100

repl = counts[counts <= threshold].index
```

```python
pd.get_dummies(df['brand'].replace(repl,'uncommon')).sample(5)
```
